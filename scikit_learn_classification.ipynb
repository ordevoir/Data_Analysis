## 2. Необходимая теоретическая база

### Классификация

Классификация — это задача обучения с учителем, в которой модель получает признаки объекта `X` и должна предсказать метку класса `y`.

Например:

- признаки: координаты точки на плоскости;
- целевая переменная: к какому из двух классов относится точка.

### Обучающая и тестовая выборки

Чтобы оценить качество модели честно, данные делят на две части:

- `train` — данные для обучения модели;
- `test` — данные для проверки модели на объектах, которые она не видела при обучении.

### Масштабирование признаков

Некоторые алгоритмы чувствительны к масштабу признаков. Для логистической регрессии и метода ближайших соседей обычно полезно использовать `StandardScaler`, который приводит признаки к сопоставимому масштабу.

### Метрики качества

Для сравнения моделей используйте не только `accuracy`, но и дополнительные метрики:

- `accuracy` — доля верных ответов;
- `precision` — насколько точны предсказания положительного класса;
- `recall` — какую долю положительного класса модель нашла;
- `f1-score` — баланс между `precision` и `recall`.

Также полезно посмотреть на `confusion_matrix`, чтобы увидеть типы ошибок модели.


## 3. Импорт библиотек и генерация данных

В этой работе используется искусственный набор данных `make_moons`. Он удобен для учебной задачи: классы видны на графике, но граница между ними нелинейная.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons

RANDOM_STATE = 42

X, y = make_moons(
    n_samples=400,
    noise=0.28,
    random_state=RANDOM_STATE
)

data = pd.DataFrame(X, columns=["feature_1", "feature_2"])
data["target"] = y

data.head()
```

## 4. Визуализация данных

Перед обучением модели полезно посмотреть на данные. Так можно заранее предположить, будет ли линейной модели достаточно гибкости для хорошего разделения классов.


```python
plt.figure(figsize=(7, 5))

scatter = plt.scatter(
    data["feature_1"],
    data["feature_2"],
    c=data["target"],
    cmap="coolwarm",
    edgecolor="k",
    alpha=0.8
)

plt.xlabel("feature_1")
plt.ylabel("feature_2")
plt.title("Синтетические данные для бинарной классификации")
plt.legend(*scatter.legend_elements(), title="Класс")
plt.grid(alpha=0.25)
plt.show()
```
